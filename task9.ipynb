```python
num_cols = data.select_dtypes(include=np.number).columns

for col in num_cols:
    data[col] = data[col].fillna(data[col].median())


cat_cols = data.select_dtypes(include='object').columns

for col in cat_cols:
    data[col] = data[col].fillna(data[col].mode()[0])


print(data.isnull().sum())
```

```python
data_encoded = pd.get_dummies(data, columns=['hotel', 'meal'])

print(data_encoded.head())
```

```python
num_data = data[['lead_time', 'adr', 'adults',
               'stays_in_weekend_nights',
               'stays_in_week_nights']]

print(num_data.head())
```

```python
num_data.hist(figsize=(12, 8), bins=20)

plt.suptitle("Histogram of Numerical Columns")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=num_data)

plt.title("Box Plot of Numerical Columns")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Skewness:")
print(num_data.skew())
```

```python
print("Kurtosis:")
print(num_data.kurtosis())
```

```python
for col in num_data.columns:

    skew = num_data[col].skew()

    if skew > 0.5:
        print(col, "- Positively Skewed")

    elif skew < -0.5:
        print(col, "- Negatively Skewed")

    else:
        print(col, "- Approximately Normal")
```
